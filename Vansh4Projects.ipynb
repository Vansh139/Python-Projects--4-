**#Project 01 - Student Management System**
Suggested features -
add students
display students
search students
update student records
delete students
calculate average marks
save records

```python
from abc import ABC, abstractmethod

# ABSTRACT BASE CLASS

class Person(ABC):

    def __init__(self, name):
        self.__name = name

    # Getter

    def get_name(self):
        return self.__name

    # Setter

    def set_name(self, name):

        if not name.strip():
            raise ValueError("Name cannot be empty.")

        self.__name = name

    # Abstract Method

    @abstractmethod
    def display_details(self):
        pass

# DERIVED CLASS

class Student(Person):

    def __init__(self, roll_number, name, marks):

        super().__init__(name)

        self.__roll_number = roll_number
        self.__marks = marks

    # GETTERS

    def get_roll_number(self):
        return self.__roll_number

    def get_marks(self):
        return self.__marks

    # SETTERS

    def set_marks(self, marks):

        if not marks:
            raise ValueError("Marks cannot be empty.")

        for mark in marks:

            if mark < 0 or mark > 100:
                raise ValueError(
                    "Marks must be between 0 and 100."
                )

        self.__marks = marks

    # CALCULATE AVERAGE

    def calculate_average(self):

        if not self.__marks:
            return 0

        return sum(self.__marks) / len(self.__marks)

    # DISPLAY DETAILS

    def display_details(self):

        print("\n------------------------------")
        print("Roll Number :", self.__roll_number)
        print("Name        :", self.get_name())
        print("Marks       :", self.__marks)
        print(
            f"Average     : {self.calculate_average():.2f}"
        )
        print("------------------------------")

# STUDENT MANAGEMENT SYSTEM

class StudentManagementSystem:

    def __init__(self):

        self.__students = []

    # FIND STUDENT

    def find_student(self, roll_number):

        for student in self.__students:

            if student.get_roll_number() == roll_number:
                return student

        return None

    # ADD STUDENT

    def add_student(self):

        try:

            roll_number = int(
                input("Enter roll number: ")
            )

            # Check duplicate roll number
            if self.find_student(roll_number) is not None:

                print("Roll number already exists.")
                return

            name = input("Enter student name: ")

            if not name.strip():

                print("Name cannot be empty.")
                return

            marks = []

            number_of_subjects = int(
                input("Enter number of subjects: ")
            )

            if number_of_subjects <= 0:

                print(
                    "Number of subjects must be greater than 0."
                )
                return

            for i in range(number_of_subjects):

                mark = float(
                    input(
                        f"Enter marks for subject {i + 1}: "
                    )
                )

                if mark < 0 or mark > 100:

                    print(
                        "Marks must be between 0 and 100."
                    )
                    return

                marks.append(mark)

            student = Student(
                roll_number,
                name,
                marks
            )

            self.__students.append(student)

            print("Student added successfully.")

        except ValueError:

            print(
                "Please enter valid numeric values."
            )

    # DISPLAY ALL STUDENTS

    def display_students(self):

        if not self.__students:

            print("No students found.")
            return

        print("\n========== ALL STUDENTS ==========")

        for student in self.__students:

            student.display_details()

    # SEARCH STUDENT

    def search_student(self):

        try:

            roll_number = int(
                input("Enter roll number to search: ")
            )

            student = self.find_student(roll_number)

            if student is None:

                print("Student not found.")
                return

            print("\nStudent Found:")
            student.display_details()

        except ValueError:

            print("Roll number must be an integer.")

    # UPDATE STUDENT

    def update_student(self):

        try:

            roll_number = int(
                input("Enter roll number to update: ")
            )

            student = self.find_student(roll_number)

            if student is None:

                print("Student not found.")
                return

            print("\n1. Update Name")
            print("2. Update Marks")

            choice = int(
                input("Enter your choice: ")
            )

            # Update Name

            if choice == 1:

                new_name = input(
                    "Enter new name: "
                )

                student.set_name(new_name)

                print(
                    "Student name updated successfully."
                )

            # Update Marks

            elif choice == 2:

                number_of_subjects = int(
                    input(
                        "Enter number of subjects: "
                    )
                )

                if number_of_subjects <= 0:

                    print(
                        "Number of subjects must be "
                        "greater than 0."
                    )
                    return

                marks = []

                for i in range(number_of_subjects):

                    mark = float(
                        input(
                            f"Enter marks for subject "
                            f"{i + 1}: "
                        )
                    )

                    if mark < 0 or mark > 100:

                        print(
                            "Marks must be between "
                            "0 and 100."
                        )
                        return

                    marks.append(mark)

                student.set_marks(marks)

                print(
                    "Student marks updated successfully."
                )

            else:

                print("Invalid choice.")

        except ValueError as e:

            print("Error:", e)

    # DELETE STUDENT

    def delete_student(self):

        try:

            roll_number = int(
                input("Enter roll number to delete: ")
            )

            student = self.find_student(roll_number)

            if student is None:

                print("Student not found.")
                return

            self.__students.remove(student)

            print(
                "Student deleted successfully."
            )

        except ValueError:

            print("Roll number must be an integer.")

    # CALCULATE AVERAGE

    def student_average(self):

        try:

            roll_number = int(
                input("Enter roll number: ")
            )

            student = self.find_student(roll_number)

            if student is None:

                print("Student not found.")
                return

            average = student.calculate_average()

            print(
                f"Average marks: {average:.2f}"
            )

        except ValueError:

            print("Roll number must be an integer.")

    # FILE HANDLING - SAVE

    def save_data(self):

        try:

            with open(
                "students.txt",
                "w"
            ) as file:

                for student in self.__students:

                    marks = ",".join(
                        str(mark)
                        for mark in student.get_marks()
                    )

                    file.write(
                        f"{student.get_roll_number()}|"
                        f"{student.get_name()}|"
                        f"{marks}\n"
                    )

            print(
                "Student data saved successfully."
            )

        except OSError as e:

            print("File error:", e)

    # FILE HANDLING - LOAD

    def load_data(self):

        try:

            with open(
                "students.txt",
                "r"
            ) as file:

                for line in file:

                    line = line.strip()

                    if not line:
                        continue

                    parts = line.split("|")

                    if len(parts) != 3:
                        continue

                    roll_number = int(parts[0])
                    name = parts[1]

                    marks = [
                        float(mark)
                        for mark in parts[2].split(",")
                    ]

                    student = Student(
                        roll_number,
                        name,
                        marks
                    )

                    self.__students.append(student)

            print(
                "Student data loaded successfully."
            )

        except FileNotFoundError:

            # First time running the program
            pass

        except (ValueError, IndexError) as e:

            print(
                "Error loading student data:",
                e
            )

        except OSError as e:

            print(
                "File error:",
                e
            )

    # MAIN MENU

    def run(self):

        self.load_data()

        while True:

            print("\n========================================")
            print("       STUDENT MANAGEMENT SYSTEM")
            print("========================================")

            print("1. Add Student")
            print("2. Display All Students")
            print("3. Search Student")
            print("4. Update Student")
            print("5. Delete Student")
            print("6. Calculate Average")
            print("7. Save Data")
            print("8. Exit")

            print("========================================")

            try:

                choice = int(
                    input("Enter your choice: ")
                )

                if choice == 1:

                    self.add_student()

                elif choice == 2:

                    self.display_students()

                elif choice == 3:

                    self.search_student()

                elif choice == 4:

                    self.update_student()

                elif choice == 5:

                    self.delete_student()

                elif choice == 6:

                    self.student_average()

                elif choice == 7:

                    self.save_data()

                elif choice == 8:

                    self.save_data()

                    print("Program closed.")
                    break

                else:

                    print(
                        "Invalid choice. "
                        "Please select 1-8."
                    )

            except ValueError:

                print(
                    "Please enter a valid menu number."
                )

system = StudentManagementSystem()
system.run()
```

**#Project 2 - Library Management System**

```python
from abc import ABC, abstractmethod

# ABSTRACT BASE CLASS

class LibraryItem(ABC):

    def __init__(self, item_id, title):
        self.__item_id = item_id
        self.__title = title

    # Getters

    def get_item_id(self):
        return self.__item_id

    def get_title(self):
        return self.__title

    # Setter

    def set_title(self, title):

        if not title.strip():
            raise ValueError("Title cannot be empty.")

        self.__title = title

    # Abstract method

    @abstractmethod
    def display_details(self):
        pass

# DERIVED CLASS

class Book(LibraryItem):

    def __init__(
        self,
        book_id,
        title,
        author,
        status="Available"
    ):

        super().__init__(book_id, title)

        self.__author = author
        self.__status = status

    # Getters

    def get_author(self):
        return self.__author

    def get_status(self):
        return self.__status

    # Setters

    def set_author(self, author):

        if not author.strip():
            raise ValueError(
                "Author name cannot be empty."
            )

        self.__author = author

    def set_status(self, status):

        if status not in ["Available", "Issued"]:
            raise ValueError(
                "Invalid book status."
            )

        self.__status = status

    # Issue Book

    def issue(self):

        if self.__status == "Issued":
            raise ValueError(
                "Book is already issued."
            )

        self.__status = "Issued"

    # Return Book

    def return_book(self):

        if self.__status == "Available":
            raise ValueError(
                "Book is already available."
            )

        self.__status = "Available"

    # Display Details

    def display_details(self):

        print("\n------------------------------")
        print("Book ID :", self.get_item_id())
        print("Title   :", self.get_title())
        print("Author  :", self.get_author())
        print("Status  :", self.get_status())
        print("------------------------------")

# LIBRARY MANAGEMENT CLASS

class Library:

    def __init__(self):

        self.__books = []

    # FIND BOOK

    def find_book(self, book_id):

        for book in self.__books:

            if book.get_item_id() == book_id:
                return book

        return None

    # ADD BOOK

    def add_book(self):

        try:

            book_id = int(
                input("Enter book ID: ")
            )

            if self.find_book(book_id) is not None:

                print(
                    "Book ID already exists."
                )
                return

            title = input(
                "Enter book title: "
            )

            if not title.strip():

                print(
                    "Book title cannot be empty."
                )
                return

            author = input(
                "Enter author name: "
            )

            if not author.strip():

                print(
                    "Author name cannot be empty."
                )
                return

            book = Book(
                book_id,
                title,
                author
            )

            self.__books.append(book)

            print(
                "Book added successfully."
            )

        except ValueError as e:

            print("Error:", e)

    # SEARCH BOOKS

    def search_books(self):

        keyword = input(
            "Enter title or author to search: "
        ).strip().lower()

        if not keyword:

            print(
                "Search keyword cannot be empty."
            )
            return

        found = False

        for book in self.__books:

            title = book.get_title().lower()
            author = book.get_author().lower()

            if keyword in title or keyword in author:

                book.display_details()
                found = True

        if not found:

            print(
                "No matching books found."
            )

    # ISSUE BOOK

    def issue_book(self):

        try:

            book_id = int(
                input("Enter book ID to issue: ")
            )

            book = self.find_book(book_id)

            if book is None:

                print("Book not found.")
                return

            book.issue()

            print(
                "Book issued successfully."
            )

        except ValueError as e:

            print("Error:", e)

    # RETURN BOOK

    def return_book(self):

        try:

            book_id = int(
                input("Enter book ID to return: ")
            )

            book = self.find_book(book_id)

            if book is None:

                print("Book not found.")
                return

            book.return_book()

            print(
                "Book returned successfully."
            )

        except ValueError as e:

            print("Error:", e)

    # DISPLAY AVAILABLE BOOKS

    def display_available_books(self):

        available_books = [

            book
            for book in self.__books
            if book.get_status() == "Available"

        ]

        if not available_books:

            print(
                "No books are currently available."
            )
            return

        print(
            "\n========== AVAILABLE BOOKS =========="
        )

        for book in available_books:

            book.display_details()

    # DISPLAY ALL BOOKS

    def display_all_books(self):

        if not self.__books:

            print("No books found.")
            return

        print(
            "\n========== ALL BOOKS =========="
        )

        for book in self.__books:

            book.display_details()

    # FILE HANDLING - SAVE

    def save_data(self):

        try:

            with open(
                "library.txt",
                "w"
            ) as file:

                for book in self.__books:

                    file.write(
                        f"{book.get_item_id()}|"
                        f"{book.get_title()}|"
                        f"{book.get_author()}|"
                        f"{book.get_status()}\n"
                    )

            print(
                "Library data saved successfully."
            )

        except OSError as e:

            print(
                "File error:",
                e
            )

    # FILE HANDLING - LOAD

    def load_data(self):

        try:

            with open(
                "library.txt",
                "r"
            ) as file:

                for line in file:

                    line = line.strip()

                    if not line:
                        continue

                    parts = line.split("|")

                    if len(parts) != 4:
                        continue

                    book_id = int(parts[0])
                    title = parts[1]
                    author = parts[2]
                    status = parts[3]

                    book = Book(
                        book_id,
                        title,
                        author,
                        status
                    )

                    self.__books.append(book)

            print(
                "Library data loaded successfully."
            )

        except FileNotFoundError:

            # First time running the program.
            pass

        except (ValueError, IndexError) as e:

            print(
                "Error loading library data:",
                e
            )

        except OSError as e:

            print(
                "File error:",
                e
            )
    # MAIN MENU

    def run(self):

        self.load_data()

        while True:

            print("\n========================================")
            print("        LIBRARY MANAGEMENT SYSTEM")
            print("========================================")

            print("1. Add Book")
            print("2. Search Books")
            print("3. Issue Book")
            print("4. Return Book")
            print("5. Display Available Books")
            print("6. Display All Books")
            print("7. Save Data")
            print("8. Exit")

            print("========================================")

            try:

                choice = int(
                    input("Enter your choice: ")
                )

                if choice == 1:

                    self.add_book()

                elif choice == 2:

                    self.search_books()

                elif choice == 3:

                    self.issue_book()

                elif choice == 4:

                    self.return_book()

                elif choice == 5:

                    self.display_available_books()

                elif choice == 6:

                    self.display_all_books()

                elif choice == 7:

                    self.save_data()

                elif choice == 8:

                    self.save_data()

                    print(
                        "Program closed."
                    )

                    break

                else:

                    print(
                        "Invalid choice. "
                        "Please select 1-8."
                    )

            except ValueError:

                print(
                    "Please enter a valid menu number."
                )

library = Library()
library.run()
```

**#Project 03 - Bank Account Management System**
suggested feature -

deposit money
withdraw money

Maintain transaction history

```python
from abc import ABC, abstractmethod

# ABSTRACT BASE CLASS

class Account(ABC):

    def __init__(self, account_number, name, balance=0):
        self.__account_number = account_number
        self.__name = name
        self.__balance = balance
        self.__transactions = []

    # GETTERS

    def get_account_number(self):
        return self.__account_number

    def get_name(self):
        return self.__name

    def get_balance(self):
        return self.__balance

    def get_transactions(self):
        return self.__transactions

    # SETTERS

    def set_name(self, name):

        if not name.strip():
            raise ValueError("Name cannot be empty.")

        self.__name = name

    # DEPOSIT

    def deposit(self, amount):

        if amount <= 0:
            raise ValueError(
                "Deposit amount must be greater than zero."
            )

        self.__balance += amount

        self.__transactions.append(
            f"Deposited: ₹{amount:.2f}"
        )

    # PROTECTED METHODS
    # Used by the derived class

    def _get_balance(self):
        return self.__balance

    def _set_balance(self, balance):
        self.__balance = balance

    def _add_transaction(self, transaction):
        self.__transactions.append(transaction)

    # ABSTRACT METHOD

    @abstractmethod
    def withdraw(self, amount):
        pass

# DERIVED CLASS

class SavingsAccount(Account):

    def withdraw(self, amount):

        if amount <= 0:
            raise ValueError(
                "Withdrawal amount must be greater than zero."
            )

        if amount > self._get_balance():
            raise ValueError(
                "Insufficient balance."
            )

        self._set_balance(
            self._get_balance() - amount
        )

        self._add_transaction(
            f"Withdrawn: ₹{amount:.2f}"
        )

# BANK MANAGEMENT CLASS

class Bank:

    def __init__(self):

        self.__accounts = []

    # FIND ACCOUNT

    def find_account(self, account_number):

        for account in self.__accounts:

            if account.get_account_number() == account_number:
                return account

        return None

    # CREATE ACCOUNT

    def create_account(self):

        try:

            account_number = int(
                input("Enter account number: ")
            )

            if self.find_account(account_number) is not None:

                print(
                    "Account number already exists."
                )
                return

            name = input(
                "Enter account holder name: "
            )

            if not name.strip():

                print(
                    "Name cannot be empty."
                )
                return

            initial_deposit = float(
                input("Enter initial deposit: ")
            )

            if initial_deposit < 0:

                print(
                    "Initial deposit cannot be negative."
                )
                return

            account = SavingsAccount(
                account_number,
                name,
                initial_deposit
            )

            if initial_deposit > 0:

                account.get_transactions().append(
                    f"Account opened with deposit: "
                    f"₹{initial_deposit:.2f}"
                )

            else:

                account.get_transactions().append(
                    "Account opened with balance: ₹0.00"
                )

            self.__accounts.append(account)

            print(
                "Account created successfully."
            )

        except ValueError as e:

            print("Error:", e)

    # DEPOSIT MONEY

    def deposit_money(self):

        try:

            account_number = int(
                input("Enter account number: ")
            )

            account = self.find_account(account_number)

            if account is None:

                print("Account not found.")
                return

            amount = float(
                input("Enter amount to deposit: ")
            )

            account.deposit(amount)

            print(
                "Money deposited successfully."
            )

            print(
                f"New balance: "
                f"₹{account.get_balance():.2f}"
            )

        except ValueError as e:

            print("Error:", e)

    # WITHDRAW MONEY

    def withdraw_money(self):

        try:

            account_number = int(
                input("Enter account number: ")
            )

            account = self.find_account(account_number)

            if account is None:

                print("Account not found.")
                return

            amount = float(
                input("Enter amount to withdraw: ")
            )

            account.withdraw(amount)

            print(
                "Money withdrawn successfully."
            )

            print(
                f"Remaining balance: "
                f"₹{account.get_balance():.2f}"
            )

        except ValueError as e:

            print("Error:", e)

    # CHECK BALANCE

    def check_balance(self):

        try:

            account_number = int(
                input("Enter account number: ")
            )

            account = self.find_account(account_number)

            if account is None:

                print("Account not found.")
                return

            print(
                "\n========== ACCOUNT DETAILS =========="
            )

            print(
                "Account Number :",
                account.get_account_number()
            )

            print(
                "Account Holder :",
                account.get_name()
            )

            print(
                f"Balance        : "
                f"₹{account.get_balance():.2f}"
            )

        except ValueError:

            print(
                "Account number must be an integer."
            )

    # TRANSACTION HISTORY

    def transaction_history(self):

        try:

            account_number = int(
                input("Enter account number: ")
            )

            account = self.find_account(account_number)

            if account is None:

                print("Account not found.")
                return

            print(
                "\n========== TRANSACTION HISTORY =========="
            )

            print(
                "Account Number :",
                account.get_account_number()
            )

            print(
                "Account Holder :",
                account.get_name()
            )

            print("-----------------------------------------")

            for transaction in account.get_transactions():

                print(transaction)

            print("-----------------------------------------")

            print(
                f"Current Balance: "
                f"₹{account.get_balance():.2f}"
            )

        except ValueError:

            print(
                "Account number must be an integer."
            )

    # DISPLAY ALL ACCOUNTS

    def display_all_accounts(self):

        if not self.__accounts:

            print("No accounts found.")
            return

        print(
            "\n========== ALL ACCOUNTS =========="
        )

        for account in self.__accounts:

            print(
                "Account Number :",
                account.get_account_number()
            )

            print(
                "Account Holder :",
                account.get_name()
            )

            print(
                f"Balance        : "
                f"₹{account.get_balance():.2f}"
            )

            print("----------------------------------")

    # SAVE DATA TO FILE

    def save_data(self):

        try:

            with open(
                "bank_accounts.txt",
                "w"
            ) as file:

                for account in self.__accounts:

                    transactions = ";".join(
                        account.get_transactions()
                    )

                    file.write(
                        f"{account.get_account_number()}|"
                        f"{account.get_name()}|"
                        f"{account.get_balance()}|"
                        f"{transactions}\n"
                    )

            print(
                "Bank data saved successfully."
            )

        except OSError as e:

            print(
                "File error:",
                e
            )

    # LOAD DATA FROM FILE

    def load_data(self):

        try:

            with open(
                "bank_accounts.txt",
                "r"
            ) as file:

                for line in file:

                    line = line.strip()

                    if not line:
                        continue

                    parts = line.split("|")

                    if len(parts) < 3:
                        continue

                    account_number = int(parts[0])
                    name = parts[1]
                    balance = float(parts[2])

                    account = SavingsAccount(
                        account_number,
                        name,
                        balance
                    )

                    if len(parts) > 3 and parts[3]:

                        transactions = parts[3].split(";")

                        for transaction in transactions:

                            account.get_transactions().append(
                                transaction
                            )

                    self.__accounts.append(account)

            print(
                "Bank data loaded successfully."
            )

        except FileNotFoundError:

            # First run — file doesn't exist yet.
            pass

        except (ValueError, IndexError) as e:

            print(
                "Error loading bank data:",
                e
            )

        except OSError as e:

            print(
                "File error:",
                e
            )

    # MAIN MENU

    def run(self):

        self.load_data()

        while True:

            print("\n========================================")
            print("       BANK ACCOUNT MANAGEMENT SYSTEM")
            print("========================================")

            print("1. Create Account")
            print("2. Deposit Money")
            print("3. Withdraw Money")
            print("4. Check Balance")
            print("5. Transaction History")
            print("6. Display All Accounts")
            print("7. Save Data")
            print("8. Exit")

            print("========================================")

            try:

                choice = int(
                    input("Enter your choice: ")
                )

                if choice == 1:

                    self.create_account()

                elif choice == 2:

                    self.deposit_money()

                elif choice == 3:

                    self.withdraw_money()

                elif choice == 4:

                    self.check_balance()

                elif choice == 5:

                    self.transaction_history()

                elif choice == 6:

                    self.display_all_accounts()

                elif choice == 7:

                    self.save_data()

                elif choice == 8:

                    self.save_data()

                    print(
                        "Program closed."
                    )

                    break

                else:

                    print(
                        "Invalid choice. "
                        "Please select 1-8."
                    )

            except ValueError:

                print(
                    "Please enter a valid menu number."
                )

bank = Bank()
bank.run()
```

**#Project 4 - Course Registration System**
suggested features- add students add courses register student for courses
drop courses
display registrations

```python
from abc import ABC, abstractmethod

# ABSTRACT BASE CLASS

class Person(ABC):

    def __init__(self, person_id, name):

        self.__person_id = person_id
        self.__name = name

    # GETTERS

    def get_person_id(self):
        return self.__person_id

    def get_name(self):
        return self.__name

    # SETTER

    def set_name(self, name):

        if not name.strip():
            raise ValueError("Name cannot be empty.")

        self.__name = name

    # ABSTRACT METHOD

    @abstractmethod
    def display_details(self):
        pass

# STUDENT CLASS - DERIVED CLASS

class Student(Person):

    def __init__(self, student_id, name):

        super().__init__(student_id, name)

        self.__registered_courses = []

    # GETTER

    def get_registered_courses(self):
        return self.__registered_courses

    # REGISTER COURSE

    def register_course(self, course_code):

        if course_code in self.__registered_courses:

            raise ValueError(
                "Student is already registered for this course."
            )

        self.__registered_courses.append(course_code)

    # DROP COURSE

    def drop_course(self, course_code):

        if course_code not in self.__registered_courses:

            raise ValueError(
                "Student is not registered for this course."
            )

        self.__registered_courses.remove(course_code)

    # DISPLAY DETAILS

    def display_details(self):

        print("\n------------------------------")
        print("Student ID :", self.get_person_id())
        print("Name       :", self.get_name())

        if self.__registered_courses:

            print(
                "Courses    :",
                ", ".join(self.__registered_courses)
            )

        else:

            print("Courses    : None")

        print("------------------------------")

# COURSE CLASS

class Course:

    def __init__(
        self,
        course_code,
        course_name,
        credits
    ):

        self.__course_code = course_code
        self.__course_name = course_name
        self.__credits = credits

    # GETTERS

    def get_course_code(self):
        return self.__course_code

    def get_course_name(self):
        return self.__course_name

    def get_credits(self):
        return self.__credits

    # SETTERS

    def set_course_name(self, course_name):

        if not course_name.strip():

            raise ValueError(
                "Course name cannot be empty."
            )

        self.__course_name = course_name

    def set_credits(self, credits):

        if credits <= 0:

            raise ValueError(
                "Credits must be greater than zero."
            )

        self.__credits = credits

    # DISPLAY COURSE

    def display_details(self):

        print("\n------------------------------")
        print(
            "Course Code :",
            self.__course_code
        )
        print(
            "Course Name :",
            self.__course_name
        )
        print(
            "Credits     :",
            self.__credits
        )
        print("------------------------------")

# COURSE REGISTRATION SYSTEM

class CourseRegistrationSystem:

    def __init__(self):

        self.__students = []
        self.__courses = []

    # FIND STUDENT

    def find_student(self, student_id):

        for student in self.__students:

            if student.get_person_id() == student_id:
                return student

        return None

    # FIND COURSE

    def find_course(self, course_code):

        for course in self.__courses:

            if course.get_course_code() == course_code:
                return course

        return None

    # ADD STUDENT

    def add_student(self):

        try:

            student_id = int(
                input("Enter student ID: ")
            )

            if self.find_student(student_id) is not None:

                print(
                    "Student ID already exists."
                )
                return

            name = input(
                "Enter student name: "
            )

            if not name.strip():

                print(
                    "Name cannot be empty."
                )
                return

            student = Student(
                student_id,
                name
            )

            self.__students.append(student)

            print(
                "Student added successfully."
            )

        except ValueError:

            print(
                "Student ID must be an integer."
            )

    # ADD COURSE

    def add_course(self):

        try:

            course_code = input(
                "Enter course code: "
            ).strip().upper()

            if not course_code:

                print(
                    "Course code cannot be empty."
                )
                return

            if self.find_course(course_code) is not None:

                print(
                    "Course code already exists."
                )
                return

            course_name = input(
                "Enter course name: "
            )

            if not course_name.strip():

                print(
                    "Course name cannot be empty."
                )
                return

            credits = int(
                input("Enter course credits: ")
            )

            if credits <= 0:

                print(
                    "Credits must be greater than zero."
                )
                return

            course = Course(
                course_code,
                course_name,
                credits
            )

            self.__courses.append(course)

            print(
                "Course added successfully."
            )

        except ValueError:

            print(
                "Credits must be a valid integer."
            )

    # REGISTER STUDENT FOR COURSE

    def register_student(self):

        try:

            student_id = int(
                input("Enter student ID: ")
            )

            student = self.find_student(student_id)

            if student is None:

                print(
                    "Student not found."
                )
                return

            course_code = input(
                "Enter course code: "
            ).strip().upper()

            course = self.find_course(course_code)

            if course is None:

                print(
                    "Course not found."
                )
                return

            student.register_course(
                course.get_course_code()
            )

            print(
                "Student registered successfully."
            )

        except ValueError as e:

            print("Error:", e)

    # DROP COURSE

    def drop_course(self):

        try:

            student_id = int(
                input("Enter student ID: ")
            )

            student = self.find_student(student_id)

            if student is None:

                print(
                    "Student not found."
                )
                return

            course_code = input(
                "Enter course code: "
            ).strip().upper()

            course = self.find_course(course_code)

            if course is None:

                print(
                    "Course not found."
                )
                return

            student.drop_course(
                course.get_course_code()
            )

            print(
                "Course dropped successfully."
            )

        except ValueError as e:

            print("Error:", e)

    # DISPLAY STUDENTS

    def display_students(self):

        if not self.__students:

            print(
                "No students found."
            )
            return

        print(
            "\n========== ALL STUDENTS =========="
        )

        for student in self.__students:

            student.display_details()

    # DISPLAY COURSES

    def display_courses(self):

        if not self.__courses:

            print(
                "No courses found."
            )
            return

        print(
            "\n========== ALL COURSES =========="
        )

        for course in self.__courses:

            course.display_details()

    # DISPLAY REGISTRATIONS

    def display_registrations(self):

        if not self.__students:

            print(
                "No students found."
            )
            return

        print(
            "\n========== COURSE REGISTRATIONS =========="
        )

        for student in self.__students:

            print(
                "\nStudent ID :",
                student.get_person_id()
            )

            print(
                "Name       :",
                student.get_name()
            )

            registered_courses = (
                student.get_registered_courses()
            )

            if not registered_courses:

                print(
                    "Courses    : None"
                )

            else:

                print(
                    "Courses    :",
                    ", ".join(registered_courses)
                )

            print("------------------------------------------")

    # SAVE DATA

    def save_data(self):

        try:

            with open(
                "course_registration.txt",
                "w"
            ) as file:

                # Save students

                for student in self.__students:

                    courses = ",".join(
                        student.get_registered_courses()
                    )

                    file.write(
                        f"STUDENT|"
                        f"{student.get_person_id()}|"
                        f"{student.get_name()}|"
                        f"{courses}\n"
                    )

                # Save courses

                for course in self.__courses:

                    file.write(
                        f"COURSE|"
                        f"{course.get_course_code()}|"
                        f"{course.get_course_name()}|"
                        f"{course.get_credits()}\n"
                    )

            print(
                "Registration data saved successfully."
            )

        except OSError as e:

            print(
                "File error:",
                e
            )

    # LOAD DATA

    def load_data(self):

        try:

            with open(
                "course_registration.txt",
                "r"
            ) as file:

                student_records = []

                for line in file:

                    line = line.strip()

                    if not line:
                        continue

                    parts = line.split("|")

                    if parts[0] == "STUDENT":

                        if len(parts) != 4:
                            continue

                        student_id = int(parts[1])
                        name = parts[2]

                        student = Student(
                            student_id,
                            name
                        )

                        if parts[3]:

                            courses = (
                                parts[3].split(",")
                            )

                            for course_code in courses:

                                student.register_course(
                                    course_code
                                )

                        self.__students.append(
                            student
                        )

                    elif parts[0] == "COURSE":

                        if len(parts) != 4:
                            continue

                        course_code = parts[1]
                        course_name = parts[2]
                        credits = int(parts[3])

                        course = Course(
                            course_code,
                            course_name,
                            credits
                        )

                        self.__courses.append(
                            course
                        )

            print(
                "Registration data loaded successfully."
            )

        except FileNotFoundError:

            # First run
            pass

        except (ValueError, IndexError) as e:

            print(
                "Error loading data:",
                e
            )

        except OSError as e:

            print(
                "File error:",
                e
            )

    # MAIN MENU

    def run(self):

        self.load_data()

        while True:

            print("\n========================================")
            print("       COURSE REGISTRATION SYSTEM")
            print("========================================")

            print("1. Add Student")
            print("2. Add Course")
            print("3. Register Student for Course")
            print("4. Drop Course")
            print("5. Display Students")
            print("6. Display Courses")
            print("7. Display Registrations")
            print("8. Save Data")
            print("9. Exit")

            print("========================================")

            try:

                choice = int(
                    input("Enter your choice: ")
                )

                if choice == 1:

                    self.add_student()

                elif choice == 2:

                    self.add_course()

                elif choice == 3:

                    self.register_student()

                elif choice == 4:

                    self.drop_course()

                elif choice == 5:

                    self.display_students()

                elif choice == 6:

                    self.display_courses()

                elif choice == 7:

                    self.display_registrations()

                elif choice == 8:

                    self.save_data()

                elif choice == 9:

                    self.save_data()

                    print(
                        "Program closed."
                    )

                    break

                else:

                    print(
                        "Invalid choice. "
                        "Please select 1-9."
                    )

            except ValueError:

                print(
                    "Please enter a valid menu number."
                )

system = CourseRegistrationSystem()
system.run()
```
